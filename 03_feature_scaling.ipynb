```python
import pandas as pd
from matplotlib import pyplot as plt

from sklearn.preprocessing import StandardScaler , MinMaxScaler, OneHotEncoder

from keras.layers import  Dense   , Input 
from keras.optimizers import Adam
from keras.losses import BinaryCrossentropy
from keras.models import Sequential
from tensorflow.random import set_seed
```

```python
x_train = pd.read_csv('./datas/Out_Stage1/x_train')
x_test  = pd.read_csv('./datas/Out_Stage1/x_test' )
x_valid = pd.read_csv('./datas/Out_Stage1/x_valid')
y_train = pd.read_csv('./datas/Out_Stage1/y_train')
y_test  = pd.read_csv('./datas/Out_Stage1/y_test' )
y_valid = pd.read_csv('./datas/Out_Stage1/y_valid')
x_train_scaled = x_train .copy()
x_test_scaled = x_test  .copy()
x_valid_scaled = x_valid .copy()
y_train_scaled = y_train .copy()
y_test_scaled = y_test  .copy()
y_valid_scaled = y_valid .copy()
```

```python
x_train.info()
```

```python

numeric = [
'AGE',
'PAY_0',
'PAY_2',
'PAY_3',
'PAY_4',
'PAY_5',
'PAY_6',
'BILL_AMT1',
'BILL_AMT2',
'BILL_AMT3',
'BILL_AMT4',
'BILL_AMT5',
'BILL_AMT6',
'PAY_AMT1',
'PAY_AMT2',
'PAY_AMT3',
'PAY_AMT4',
'PAY_AMT5',
'PAY_AMT6'
]
for column in numeric :
    plt.close()
    plt.title(column)
    x_train[column].hist(bins=30)
    plt.show()

```

```python
standrad_cols = ['PAY_0','PAY_2','PAY_3','PAY_4','PAY_5','PAY_6',]
minmax_cols = ['AGE','BILL_AMT1','BILL_AMT2','BILL_AMT3','BILL_AMT4','BILL_AMT5','BILL_AMT6','PAY_AMT1','PAY_AMT2','PAY_AMT3','PAY_AMT4','PAY_AMT5','PAY_AMT6']
onehot_cols = ['MARRIAGE', 'EDUCATION','SEX']

minmax_scaler = MinMaxScaler()
standrad_scaler = StandardScaler()
onehot_encoder = OneHotEncoder(sparse_output=False)

x_train_scaled[standrad_cols] = standrad_scaler.fit_transform(x_train[standrad_cols])
x_test_scaled[standrad_cols] = standrad_scaler.transform(x_test[standrad_cols])
x_valid_scaled[standrad_cols] = standrad_scaler.transform(x_valid[standrad_cols])

x_train_scaled[minmax_cols] = minmax_scaler.fit_transform(x_train[minmax_cols])
x_test_scaled[minmax_cols] = minmax_scaler.transform(x_test[minmax_cols])
x_valid_scaled[minmax_cols] = minmax_scaler.transform(x_valid[minmax_cols])

x_train_encoded = onehot_encoder.fit_transform(x_train[onehot_cols],)
x_test_encoded = onehot_encoder.transform(x_test[onehot_cols])
x_valid_encoded = onehot_encoder.transform(x_valid[onehot_cols])
```

```python
x_train_encoded =pd.DataFrame(data= x_train_encoded ,columns=onehot_encoder.get_feature_names_out())
x_test_encoded =pd.DataFrame(data= x_test_encoded ,columns=onehot_encoder.get_feature_names_out())
x_valid_encoded =pd.DataFrame(data= x_valid_encoded ,columns=onehot_encoder.get_feature_names_out())
```

```python
x_train_scaled.drop(onehot_cols , axis=1 , inplace=True)
x_test_scaled.drop(onehot_cols , axis=1 , inplace=True)
x_valid_scaled.drop(onehot_cols , axis=1 , inplace=True)


x_train_scaled =pd.concat([x_train_scaled,x_train_encoded ],axis=1 )
x_test_scaled =pd.concat([x_test_scaled,x_test_encoded ],axis=1 )
x_valid_scaled =pd.concat([x_valid_scaled,x_valid_encoded ],axis=1 )
```

```python
set_seed(42)
model_simplenormal = Sequential([
    Input(shape=(33,)),
    Dense(64, activation='relu'),
    Dense(64 , activation='relu'),
    Dense(1,activation='sigmoid')
])

model_simplenormal.compile(optimizer=Adam(learning_rate=1e-3) , loss=BinaryCrossentropy() ,metrics=['accuracy'])
```

```python
history_simplenormal_model = model_simplenormal.fit(x_train_scaled , y_train_scaled  , epochs=5 , validation_data=(x_valid_scaled,y_valid_scaled) , )
```

```python
x_train_out =pd.concat([x_train_scaled,x_train[onehot_cols] ],axis=1 )
x_test_out =pd.concat([x_test_scaled,x_test[onehot_cols]],axis=1 )
x_valid_out =pd.concat([x_valid_scaled,x_valid[onehot_cols] ],axis=1 )
```

```python
x_train_out .to_csv('./datas/Out_Stage2/x_train',index=False)
x_test_out .to_csv('./datas/Out_Stage2/x_test',index=False)
x_valid_out .to_csv('./datas/Out_Stage2/x_valid',index=False)
y_train_scaled .to_csv('./datas/Out_Stage2/y_train',index=False)
y_test_scaled .to_csv('./datas/Out_Stage2/y_test',index=False)
y_valid_scaled .to_csv('./datas/Out_Stage2/y_valid',index=False)
```
