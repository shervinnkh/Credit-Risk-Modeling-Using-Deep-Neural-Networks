```python
import pandas as pd
from keras.models import load_model
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report, roc_auc_score
from ShervinKhLib import load_var
from sklearn.metrics import accuracy_score
```

```python
numeric_cols = ['PAY_0','PAY_2','PAY_3','PAY_4','PAY_5','PAY_6','AGE','BILL_AMT1','BILL_AMT2',
                'BILL_AMT3','BILL_AMT4','BILL_AMT5','BILL_AMT6','PAY_AMT1','PAY_AMT2','PAY_AMT3','PAY_AMT4','PAY_AMT5','PAY_AMT6']
x_test = pd.read_csv('./datas/Out_Stage3/x_test')
y_test = pd.read_csv('./datas/Out_Stage3/y_test')
threshold  = load_var('./datas/best_t')
```

```python
model = load_model('model_brain.h5')
```

```python
y_probs = model.predict(
    [x_test['SEX'], x_test['MARRIAGE'], x_test['EDUCATION'], x_test[numeric_cols]],
    verbose=0
).ravel()
y_pred = (y_probs >= threshold ).astype(int)

acc = accuracy_score(y_test, y_pred)
print("Model Accuracy :", round(acc * 100, 2), "%")
```

```python
y_prob = model.predict([x_test['SEX'],x_test['MARRIAGE'],x_test['EDUCATION'],x_test[numeric_cols]],verbose=0)
y_pred = (y_prob >= threshold ).astype(int)
matrix = confusion_matrix(y_test,y_pred)
print(matrix)
```

```python
y_pred = y_prob.ravel()
print(classification_report(y_test, y_pred > threshold ))
print("ROC AUC:", roc_auc_score(y_test, y_pred))
```
