```python
import pandas as pd
from sklearn.model_selection import train_test_split
from ShervinKhLib.ai_tools import train_test_valid_split
```

```python
data = pd.read_csv('./UCI_Credit_Card.csv')
```

```python
x = data.drop([data.columns[-1],'ID'],axis=1)
y = data[data.columns[-1]]
```

```python
x_train , x_test , x_valid , y_train , y_test , y_valid = train_test_valid_split(x,y,train_size=0.7 , test_size=0.15)
```

```python

x_train .to_csv('./datas/Out_Stage1/x_train',index=False)
x_test  .to_csv('./datas/Out_Stage1/x_test' ,index=False)
x_valid .to_csv('./datas/Out_Stage1/x_valid',index=False)
y_train .to_csv('./datas/Out_Stage1/y_train',index=False)
y_test  .to_csv('./datas/Out_Stage1/y_test' ,index=False)
y_valid .to_csv('./datas/Out_Stage1/y_valid',index=False)
```
