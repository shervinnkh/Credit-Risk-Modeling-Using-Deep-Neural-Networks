```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from keras.models import load_model
from sklearn.metrics import confusion_matrix
from sklearn.metrics import precision_recall_curve
from ShervinKhLib import save_var
```

```python
numeric_cols = ['PAY_0','PAY_2','PAY_3','PAY_4','PAY_5','PAY_6','AGE','BILL_AMT1','BILL_AMT2',
                'BILL_AMT3','BILL_AMT4','BILL_AMT5','BILL_AMT6','PAY_AMT1','PAY_AMT2','PAY_AMT3','PAY_AMT4','PAY_AMT5','PAY_AMT6']
x_valid = pd.read_csv('./datas/Out_Stage3/x_valid')
y_valid = pd.read_csv('./datas/Out_Stage3/y_valid')
```

```python
model = load_model('model_brain.h5')
```

```python
y_probs = model.predict(
    [x_valid['SEX'], x_valid['MARRIAGE'], x_valid['EDUCATION'], x_valid[numeric_cols]],
    verbose=0
)

ps, rs, ts = precision_recall_curve(y_valid, y_probs)

target_precision = 0.74
idxs = np.where(ps[:-1] >= target_precision)[0]
best_idx = idxs[0]
best_t = ts[best_idx]

plt.plot(ts, ps[:-1], label="precision")
plt.plot(ts, rs[:-1], label="recall")
plt.vlines(best_t, 0, 1.0, "k", "dotted", label="threshold")
plt.ylim(0, 1)
plt.legend()
plt.show()
```

```python
save_var('./datas/best_t',best_t)
```
