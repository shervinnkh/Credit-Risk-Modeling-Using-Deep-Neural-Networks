```python
import pandas as pd
from ShervinKhLib import load_var

from keras.layers import Flatten , Dense  , Embedding , Input , Concatenate
from keras.optimizers import Adam
from keras.losses import BinaryCrossentropy
from keras.models import Sequential
from tensorflow.random import set_seed
```

```python
x_train = pd.read_csv('./datas/Out_Stage1/x_train')
x_test  = pd.read_csv('./datas/Out_Stage1/x_test' )
x_valid = pd.read_csv('./datas/Out_Stage1/x_valid')
y_train = pd.read_csv('./datas/Out_Stage1/y_train')
y_test  = pd.read_csv('./datas/Out_Stage1/y_test' )
y_valid = pd.read_csv('./datas/Out_Stage1/y_valid')
```

```python
set_seed(42)
simple_model = Sequential([
    Input(shape=(23,)),
    Dense(64, activation='relu'),
    Dense(64 , activation='relu'),
    Dense(1,activation='sigmoid')
])

simple_model.compile(optimizer=Adam(learning_rate=1e-3) , loss=BinaryCrossentropy() ,metrics=['accuracy'])
```

```python
history_simple_model = simple_model.fit(x_train , y_train  , epochs=5 , validation_data=(x_valid,y_valid) , )
```
